## List Slicing  

Slicing lets you access **specific parts** of a list whether it's a single element, a range, or every few items.

Python supports slicing for any **indexable sequence**, like lists or strings.

### Slicing Syntax:

```python
list[start : end : step]
```

Or simply:

```python
list[start : end]
```

- `start` → index to begin (inclusive)  
- `end` → index to stop (exclusive)  
- `step` → how many items to skip (optional)

All parts are **optional**, and Python will use defaults if you leave them out.

---

### Example:

```python
my_list = [10, 20, 30, 40, 50]

print(my_list[1:4])     # [20, 30, 40]
print(my_list[:3])      # [10, 20, 30]
print(my_list[::2])     # [10, 30, 50]
```


The `range()` function is used to generate a sequence of numbers — often used in loops or to create lists of numbers.

range(start, stop, step)
- start (optional) → where the sequence begins (default is 0)
- stop → where the sequence ends (but not included)
- step (optional) → how much to count by (default is 1)

````python
listof10numbers = list(range(10))
````

````python
listof10numbers #notice that when you give 10 , the sequence ends at 10 but not INCLUDED 
````

````python
listof10numbers[0:10:3]
````

````python
#lets firts create a list of 20 numbers, 0 to 19
listof20numbers = list(range(20))
print(listof20numbers)
#next lets slice
print("elements from index 4 to 7:", listof20numbers[4:8])

print("even indexed values, starting at index 0:", listof20numbers[::2]) 
print("odd indexed values, starting at index 1:", listof20numbers[1::2]) 
````

    do_default_action()

##  File I/O in Python : Reading and Writing Files  
***

Python makes it easy to work with files using the built-in `open()` function. It lets you read, write, or modify text and binary files.

---

###  Key Concepts:
- Files in Python are handled through a **file object**
- Similar to `FileStream` in other programming languages
- Since a file is a system resource, it’s important to **close it after use**
- Best practice: use a **`with` statement** to manage files automatically

---

###  Example (Using `with`):
What 'with' Does:
If the file doesn’t exist → it creates a new one
If the file exists → it deletes all its contents before writing

```python

# File is not closed after
file = open("example.txt", "r")
content = file.read()
file.close()  # You must remember to close it


# File is automatically closed after this block
with open("example.txt", "r") as file:
    content = file.read()
    print(content)






````python
## lets create a file, overwite old files
with open("fileofsentences", 'w') as f:
    f.write("This line is important \n")
    f.write("So is this line \n")
    
````

````python
## lets overwite old files
with open("fileofsentences", 'w') as f:
    f.write("blahblahblah \n")
    f.write("hahahaah \n")

````

````python
# let's verify if it was really created.
# For that, let's find out which directory we're working from

import os #helps us interact with operating system

print(os.path.abspath(os.curdir)) #returns our current directory
````

````python
#use append to edit, existing contect is untouched
with open("fileofsentences", 'a') as f:
    f.write('appended new line \n')
````

## Exception Handling in Python  
***

### What is it?

Sometimes, your code might run into errors like dividing by zero or trying to open a file that doesn’t exist. These are called **exceptions**.

Python gives you a way to handle these situations so your program doesn’t crash. This is done using `try` and `except`.

---

### Basic Syntax

```python
try:
    # Code that might cause an error
except:
    # Code that runs *if* an error happens


<img src="images/exceptions.png" alt="Concept-Alert" style="width:300px; height:400px; float:left; margin-right:15px;"/>

````python
#what if we dont know what will happen
try:
    risky_code_here()
except Exception as e: #This catches all “normal” errors (everything that inherits from Exception).Good for debugging when you’re unsure which error will occur.
    print("Something went wrong:", e) 
````

````python
#shows how exception handling keeps your program running, even when something goes wrong! 
#Without try-except, the program would crash at 1 / 0 and stop completely.
print('hello world')

try:
    1 / 0
except ZeroDivisionError:
    print('exception occured')

print('bye world')
````
