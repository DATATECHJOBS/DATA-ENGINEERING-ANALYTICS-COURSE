## Function Arguments  
***

When you define a function, you can pass in **parameters** pieces of data the function can use.

---

### Key Points:

- Parameters can have **default values**  
  → Example: `def func(x, y=3)` → `y` has a default value of `3`

- Arguments can be passed by:
  - **Position** → `func(10, 20)`
  - **Name (keyword)** → `func(x=10, y=20)`

- Use `*args` to accept **any number of extra positional arguments**  
  → Example: `def func(*args)`

- Use `**kwargs` to accept **any number of extra keyword arguments**  
  → Example: `def func(**kwargs)`

- In Python, arguments are passed by **object reference** (not copied). This means if you change a mutable object (like a list), the change will affect the original.

---

Let’s look at some examples!

## Lambda Functions  
***

Lambda functions are one of Python’s most powerful and sometimes confusing features for beginners.

They allow you to create **small, anonymous functions** without using the `def` keyword.

---

### What does "anonymous" mean?  
It just means the function **has no name** you define and use it on the spot.

Lambda functions behave just like normal functions created with `def`, but they are meant for **quick, short tasks**.

---

### Key Differences:
- A lambda can only have **one expression** (no multiple lines or complex logic)
- Mostly used when you need to **pass a simple function as an argument**

---

### Syntax:
```python
lambda arguments: expression


````python
def double(number):
    result = number*2
    return result
double(2)
````

We can simplify it into a lambda expression shown below:

````python
double_lambda = lambda number: number*2
````

````python
double_lambda(2)
````

````python
add_lambda = lambda a,b: a+b

add_lambda(2,2)
````

# Example: Dog Class with Attributes and Methods  
***

Let's create a simple class for a **Dog** that has attributes like `name` and `breed`, and a method called `.bark()` that prints a message.

Why Attributes Matter:
They hold data specific to each object
They make your objects unique (each dog can have a different name)
They let you use that data in methods or later in your program



### 1. first we create a class and Name

### 2. `__init__` Components (if we want attributes)

- `__init__` → Special method that runs when an object is created. It sets up the object or initializes the attributes of an object
- `self` → Refers to the current/'this' object being created 
- `name` and `breed` → Inputs (parameters) you pass in when creating the dog.
- `self.name = name` → Stores the value as an **attribute** inside the object so you can access it later.


````python
class Dog:
    def __init__(self, name, breed):
        # These are attributes
        #They belong to the object (instance), not the class itself
        self.name = name
        self.breed = breed

    def bark(self):
        # This is a method/ function
        print(f"{self.name} says woof!")

````

````python
# Creating an instance of Dog
#we **instantiate** the Dog class
my_dog = Dog('Buddy','Golden Retriever')
#my_dog = Dog(breed = 'Golden Retriever', name = 'Buddy')
````

````python
type(my_dog) #my_dog is an object of type Dog
````

````python
# Accessing/calling attributes: we use '.' to access attributes
my_dog.name  # Output: Buddy
my_dog.breed  # Output: Golden Retriever

# Calling a method
my_dog.bark()     # Output: Buddy says woof!
````

## Methods in Python Classes  
***

**Methods** are functions that are defined **inside a class**. They let objects **do things** using their own data (attributes).

---

### Why are methods useful?

- They let each object act on **its own data** e.g bark method
- They are key to **encapsulation**, meaning each object controls its own behavior(You don’t need to know how it works internally)
- They help organize code, especially in **larger programs**

You can think of a **method** as a function that belongs to an object and it always takes `self` as the first argument to refer to **that specific object**.


````python
class Circle:
    pi = 3.14 #class variable
    def __init__(self, radius =1):
        self.radius = radius #attribute / its also an instance variable
        
    def area(self):
        return self.radius*self.radius*Circle.pi
        
    def setRadius(self,radius):
        self.radius = radius
        
    
````

````python
c = Circle(4)
````

````python
type(c)
````

````python
c.radius
````

````python
c.area()
````

````python
c.setRadius(2)
````

````python
c.area()
````

## Instance Methods vs Class Methods vs Static Methods  
***

### 1. Instance Methods (`def method(self)`)

- Most common type of method  
- Can access and modify the **object's attributes** using `self`  
- Tied to a specific object (or instance) of the class  

Use when the method needs to work with **individual object data**

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def bark(self):  # Instance method
        print(f"{self.name} says woof!")


## Class Method   
***

Class methods act on the class itself, not on specific objects.
They use cls (instead of self) and are defined using the @classmethod decorator.

Use a class method when:

You need to work with class-level data (shared by all objects)
You want to create alternate constructors

```python



````python
class Product:
    tax_rate = 0.1  # Class-level attribute (shared by all products)

    @classmethod
    def change_tax_rate(cls, new_rate):
        cls.tax_rate = new_rate
````

````python
print(Product.tax_rate)  # Output: 0.1
````

````python
Product.change_tax_rate(0.2)
print(Product.tax_rate)  # we have no changed the class!
````

## Static Methods in Python  
***

### What is a Static Method?

A **static method** is a function inside a class that:

- Doesn’t need access to the object (`self`)
- Doesn’t need access to the class (`cls`)
- Acts like a **regular function**, but is placed inside a class for better organization

It's defined using the `@staticmethod` decorator.

---

### Why Use a Static Method?

Use a static method when:

- The method is **related** to the class logically
- But it doesn’t need to access or modify any object or class-level data
- You just want to **group it** with the class for structure or clarity

---



````python
class MathHelper:
    @staticmethod
    def add(x, y):
        return x + y
````

````python
print(MathHelper.add(3, 5))  # Output: 8
````

````python
class MyClass:
    def instance_method(self):
        print('instance method called', self)

    @classmethod
    def class_method(cls):
        print('class method called', cls)

    @staticmethod
    def static_method():
        print('static method called')

obj = MyClass()

obj.instance_method()
MyClass.class_method()
MyClass.static_method()
````

````python
class Employee:
    company = 'Techcorp'

    def __init__(self, name,role, salary):
        self.name = name
        self.role = role
        self.salary = salary
        
    #instance method
    def display_info(self):
        print(f"{self.name} - {self.role} - {self.salary} ")

    @classmethod
    def change_company(cls,new_name):
        cls.company = new_name

    @staticmethod
    def calculate_bonus(salary):
        return salary*0.2

    
        
````

````python
emp1 = Employee('Alice', 'Dev',100000)
````

````python
# Calling instance method
emp1.display_info()
````

````python
# Calling class method to change company name
#emp1.company
#Employee.change_company('Meta')
emp1.company
````

````python
# Now both employees will reflect the updated company
 emp1.company

````

````python
# Calling static method to calculate bonus
bonusemp1 = emp1.calculate_bonus(100000)
bonusemp1 = emp1.calculate_bonus(emp1.salary)
````
