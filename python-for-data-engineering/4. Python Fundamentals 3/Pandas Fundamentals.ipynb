## Creating Data

You can create a simple Series or DataFrame from a list or dictionary.

````python
a_list =[10, 20, 30]
````

````python
# Create a DataFrame from a dictionary
data = {'Name': ['Alice', 'Bob'], 'Age': [25, 30]}
df = pd.DataFrame(data)
print(df)
````

## Index in a Pandas Series

In a Pandas Series, each value has a corresponding index kind of like a label.

You can use these index labels (or numbers) to quickly look up values. Think of a Series like a dictionary: index → value.

````python
import pandas as pd

# Create two Series representing fruit prices in different stores
store_a = pd.Series([0.99, 1.29, 2.49], index=['Apple', 'Banana', 'Mango'])
store_b = pd.Series([1.09, 1.19, 2.79], index=['Apple', 'Banana', 'Orange'])
````

````python
print(store_a)
````

````python
print(store_a['Mango']) 
````

A pandas series is usually homogenous (meaning all the same data types), its 1 dimensional , is similar to a excel spreadsheet column

## DataFrame Basics

What is a DataFrame?
A DataFrame is a 2D table (like an Excel sheet) with labeled rows and columns. It can hold different data types like numbers, text, or booleans.

Rows = individual records
Columns = variables/features
Comes from the pandas library
Each DataFrame has:

Data
Index (row labels)
Columns (column labels)
