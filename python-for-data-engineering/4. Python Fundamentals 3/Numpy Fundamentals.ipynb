# NumPy Basics

NumPy (Numerical Python) is a powerful Python library for numerical operations. It provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on them.

## What is an ndarray?

The core of NumPy is the `ndarray` (n-dimensional array). It stores elements of the same type and allows efficient numerical computations.

- Items are indexed using zero-based indices.
- All elements share the same data type (`dtype`).
- Arrays can be 1D, 2D, 3D, and more.

## Basic Operations

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
#print(a + b) #vectorized addition, but this will fail if arrays are if diff lenght
#print(a * b)
print(a ** 2)
````

````python
#this gives an error when lengths dont match
a = np.array([1, 2, 3])
b = np.array([4, 5])
print(a + b)
````

````python
print(np.mean(a))
````

````python
print(np.sum(a))
````

````python
print(np.max(a))
````
